```python
import pandas as pd
from scipy.stats import kruskal
```

```python
df = pd.read_excel("/content/tesla_used_car_sold-2022-08.xlsx")

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
print(df["model"].value_counts())
```

```python
m3 = df[df["model"] == "m3"]["sold_price"]
ms = df[df["model"] == "ms"]["sold_price"]
mx = df[df["model"] == "mx"]["sold_price"]
my = df[df["model"] == "my"]["sold_price"]
```

```python
H, p = kruskal(m3, ms, mx, my)

print("Kruskal-Wallis H statistic:", H)
print("Degrees of freedom:", 3)
print("p-value:", p)
```

```python
print("Kruskal-Wallis Test")
print("--------------------")
print(f"H statistic: {H:.3f}")
print("Degrees of freedom: 3")
print(f"p-value: {p:.3e}")

if p < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```
